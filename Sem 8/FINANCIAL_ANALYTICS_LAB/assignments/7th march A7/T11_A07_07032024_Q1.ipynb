```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import statsmodels.api as sm
```

```python
data = pd.read_excel('/content/drive/MyDrive/dataset/Data for Day of the week effect.xlsx', sheet_name=None)
```

```python
excel={}

for sheet_name, df in data.items():
  excel[sheet_name]=df

excel['BSE']
```

```python
for sheet_name, df in data.items():
  df['Date'] = pd.to_datetime(df['Date'])
  df['Day_of_week'] = df['Date'].dt.dayofweek
  if sheet_name == "BSE":
    df['Return'] = np.log(df['Close']).diff()
  else:
    df['Return'] = np.log(df['Adj Close']).diff()
  df['Return'].fillna(0,inplace=True)
  excel[sheet_name]=df
  print("Sheet name:", sheet_name)
  print(excel[sheet_name]['Return'])
  print('\n')
```

```python
for sheet_name, df in data.items():
  df = df[['Day_of_week','Return']]
  excel[sheet_name]=df

excel['BSE']
```

```python
for sheet_name, df in data.items():
  df = df[['Day_of_week','Return']]
  days_dummies = pd.get_dummies(df['Day_of_week'], prefix='Day')
  df = df.drop(['Day_of_week'], axis=1)
  df = pd.concat([df, days_dummies], axis=1)
  excel[sheet_name]=df

excel['BSE']
```

```python
X = excel['BSE'][['Day_0','Day_1','Day_2','Day_3','Day_4']]
X = sm.add_constant(X)
y = excel['BSE']['Return']

# Fit the regression model
model = sm.OLS(y, X).fit()

# Print the regression results
print(model.summary())
```
