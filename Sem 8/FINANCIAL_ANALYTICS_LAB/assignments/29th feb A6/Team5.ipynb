```python
import pandas as pd
import statsmodels.api as sm

# Splitting the data into independent and dependent variables
X = df.drop(df.columns[0],axis=1)  # Features
y = df[df.columns[0]]  # Target variable

# Adding constant term to the independent variables (required for statsmodels)
X = sm.add_constant(X)

# Creating and fitting the model
model = sm.OLS(y, X).fit()

# Summary of the regression model
print(model.summary())
```

```python
import pandas as pd
import statsmodels.api as sm

df_2=df.drop('Amount of Prepaid Card ($)',axis=1)

# Splitting the data into independent and dependent variables
X = df_2.drop('Days per month at Starbucks', axis=1)  # Features
y = df_2['Days per month at Starbucks']  # Target variable


# Adding constant term to the independent variables (required for statsmodels)
X = sm.add_constant(X)

# Creating and fitting the model
model = sm.OLS(y, X).fit()

# Summary of the regression model
print(model.summary())
```

```python
df_2.corr()['Days per month at Starbucks']
```

For the variable "Age," the p-value is 0.445, which is greater than the common significance level of 0.05. This indicates that there is insufficient evidence to reject the null hypothesis that the coefficient for "Age" is zero.
Similarly, for "Income ($1,000)," the p-value is 0.185, indicating that changes in income are not associated with statistically significant changes in the number of days spent at Starbucks.



The only variable with a statistically significant coefficient is **"Cups of Coffee per day," with a p-value of 0.016.** However, it's important to note that this significance level might not be stringent enough depending on the context and the significance threshold chosen.

```python
import pandas as pd
import statsmodels.api as sm

pf=pf.drop(pf.columns[0],axis=1)

# Splitting the data into independent and dependent variables
X = pf.drop('REVENUE', axis=1)  # Features
y = pf['REVENUE']  # Target variable


# Adding constant term to the independent variables (required for statsmodels)
X = sm.add_constant(X)

# Creating and fitting the model
model = sm.OLS(y, X).fit()

# Summary of the regression model
print(model.summary())
```
