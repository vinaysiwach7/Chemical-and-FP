```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_excel('/content/drive/MyDrive/dataset/Data_Q1.xlsx')
```

```python
df
```

```python
X = df.iloc[:,1:5].values
y = df.iloc[:,0].values

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

r2 = r2_score(y, y_pred)
print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
print("R-squared:", r2)
```

```python
correlation_matrix = np.corrcoef(np.column_stack((X, y)), rowvar=False)
correlation_with_target = correlation_matrix[:-1, -1]
correlation_with_target
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_with_target.reshape(1, -1), annot=True, cmap='coolwarm', fmt='.2f',
            xticklabels=['Age', 'Days per month', 'Cups of Coffee','Income($1000)'], yticklabels=['Correlation with Amount of Prepaid Card'])
plt.title('Correlation with Amount of Prepaid Card')
plt.show()
```
