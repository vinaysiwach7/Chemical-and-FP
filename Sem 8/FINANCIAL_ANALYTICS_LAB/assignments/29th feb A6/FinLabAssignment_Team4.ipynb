# ***Q1***

```python
import pandas as pd
df = pd.read_csv("/content/Data_Q1 - Sheet1.csv")
```

```python
df
```

```python
import statsmodels.api as sm
from sklearn.metrics import r2_score, mean_squared_error
X = df.drop('Amount of Prepaid Card($)', axis=1).values
y = df['Amount of Prepaid Card($)'].values
```

```python
# Calculate correlation coefficients
correlation_with_target = df.corr()['Amount of Prepaid Card($)'].drop('Amount of Prepaid Card($)')

print("Correlation of features with the target variable:")
print(correlation_with_target)
```

```python
X = sm.add_constant(X)
model = sm.OLS(y, X)
results = model.fit()
print(results.summary())
```

```python
# Get R-squared value
r_squared = results.rsquared
print("R-squared:", r_squared)
col=['Age', 'Days per month at starbucks', 'Cups of Coffee per Day', 'Income ($1)']
print("The P values are")
# Get p-values for each predictor
p_values = results.pvalues[1:]  # Drop the constant term
for a,b in zip(col ,p_values):
  print(a,b)
print("The coefficients are")
# Get coefficients for each predictor
coefficients = results.params[1:]  # Drop the constant term
for a,b in zip(col ,coefficients):
  print(a,b)

# Identify significant predictors based on p-values (common threshold is 0.05)
significant_predictors = p_values[p_values < 0.05].tolist()
```

```python
age_ranges = [(x, x+10) for x in range(df['Age'].min(), df['Age'].max(), 10)]

# 3-5. Calculate correlations for each age range
correlations = {}
for age_range in age_ranges:
    # 3. Subset data
    age_min, age_max = age_range
    subset = df[(df['Age'] >= age_min) & (df['Age'] < age_max)]

    # 4. Calculate correlation
    correlation = subset['Age'].corr(df['Amount of Prepaid Card($)'])

    # 5. Store correlation
    correlations[age_range] = correlation

# Print correlations
for age_range, correlation in correlations.items():
    print(f"Age range {age_range}: Correlation = {correlation}")
```

```python
X = df.drop(['Amount of Prepaid Card($)','Days per month at starbucks'], axis=1).values
y = df['Days per month at starbucks'].values
```

```python
# Calculate correlation coefficients
dfx = df.drop(['Amount of Prepaid Card($)'],axis=1)
correlation_with_target = dfx.corr()['Days per month at starbucks'].drop('Days per month at starbucks')

print("Correlation of features with the target variable:")
print(correlation_with_target)
```

```python
X = sm.add_constant(X)
model = sm.OLS(y, X)
results = model.fit()
print(results.summary())
```

```python
# Get R-squared value
r_squared = results.rsquared
print("R-squared:", r_squared)
col=['Age', 'Cups of Coffee per Day', 'Income ($1)']
print("The P values are")
# Get p-values for each predictor
p_values = results.pvalues[1:]  # Drop the constant term
for a,b in zip(col ,p_values):
  print(a,b)
print("The coefficients are")
# Get coefficients for each predictor
coefficients = results.params[1:]  # Drop the constant term
for a,b in zip(col ,coefficients):
  print(a,b)

# Identify significant predictors based on p-values (common threshold is 0.05)
significant_predictors = p_values[p_values < 0.05].tolist()
j=0
for i in range(len(p_values)):
    if p_values[i]==significant_predictors[j]:
      print(col[i],"is a significant predictor")
      j=j+1
      i=0
      if j==len(significant_predictors):
        break
```
