```python
from statsmodels.tsa.stattools import zivot_andrews
# Example function to test for unit root with structural break
def test_with_structural_break(array):
    result = zivot_andrews(array.flatten())
    return {"Test Statistic": result[0], "p-value": result[1], "Break Point": result[2]}

# Example usage
# Replace `Brazil` with the actual data array for Brazil
canada_result = test_with_structural_break(canada)
print("Canada:", canada_result)
france[5] = france[0]
france_result = test_with_structural_break(france)
print("france:", france_result)
germany_result = test_with_structural_break(germany)
print("germany:", germany_result)
italy_result = test_with_structural_break(italy)
print("italy:", italy_result)
japan_result = test_with_structural_break(japan)
print("japan:", japan_result)
```

```python
from statsmodels.tsa.stattools import coint
def cointegration_test(data1, data2):
  result = coint(data1, data2)
  print(f"Cointegration Statistic: {result[0]}")
  print(f"p-value: {result[1]}\n")
  if result[1] < 0.05:
      print("FDI is cointegrated\n")
  else:
      print("FDI is not cointegrated\n")
  print("-------------------------------------\n")
```

```python
canada = np.array(canada)
france = np.array(france)
germany = np.array(germany)
italy = np.array(italy)
japan = np.array(japan)
print("Cointegration Test between Canada and France")
cointegration_test(canada, france)

print("Cointegration Test between Canada and Germany")
cointegration_test(canada, germany)

print("Cointegration Test between Canada and Italy")
cointegration_test(canada, italy)

print("Cointegration Test between Canada and Japan")
cointegration_test(canada, japan)

print("Cointegration Test between France and Germany")
cointegration_test(france, germany)

print("Cointegration Test between France and Italy")
cointegration_test(france, italy)

print("Cointegration Test between France and Japan")
cointegration_test(france, japan)

print("Cointegration Test between Germany and Italy")
cointegration_test(germany, italy)

print("Cointegration Test between Germany and Japan")
cointegration_test(germany, japan)

print("Cointegration Test between Italy and Japan")
cointegration_test(italy, japan)
```
